Indexing and slicing

```python
arr = np.array([10, 20, 30, 40, 50])

# Access the first element
print(arr[0])

# Access the third element
print(arr[2])

# Access the last element
print(arr[-1])
```

```python
arr = np.array([10, 20, 30, 40, 50, 60])

# Get elements from index 1 to 3
print(arr[1:4])

# Get elements from beginning to index 3
print(arr[:4])

# Get elements from index 2 to end
print(arr[2:])

# Get every second element
print(arr[::2])

# Reverse the array
print(arr[::-1])
```

```python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

# Access first row
print(arr[0])

# Access element at row 1, column 2
print(arr[1, 2])

# Access first column
print(arr[:, 0])

# Access second column
print(arr[:, 1])

# Get first two rows
print(arr[:2, :])

# Get first two columns
print(arr[:, :2])
```

Statistical Functions

```python
marks = np.array([70, 80, 90, 60, 100])

# Calculate total
print(np.sum(marks))

# Calculate average
print(np.mean(marks))

# Find maximum value
print(np.max(marks))

# Find minimum value
print(np.min(marks))

# Find median
print(np.median(marks))

# Find standard deviation
print(np.std(marks))

# Find variance
print(np.var(marks))

# Find index of maximum value
print(np.argmax(marks))

# Find index of minimum value
print(np.argmin(marks))

# Calculate percentile
print(np.percentile(marks, 50))

# Calculate cumulative sum
print(np.cumsum(marks))

# Calculate cumulative product
print(np.cumprod(marks))
```

Sorting and seaching

```python
arr = np.array([50, 10, 40, 20, 30])

# Sort in ascending order
print(np.sort(arr))

# Get the indices that would sort the array
print(np.argsort(arr))

# Sort in descending order
print(np.sort(arr)[::-1])
```

```python
arr = np.array([10, 20, 30, 40, 50, 30])

# Find indices where values equal 30
print(np.where(arr == 30))

# Find indices where values are greater than 25
print(np.where(arr > 25))

# Return indices of non-zero elements
print(np.nonzero(arr))
```

filtering arrays

```python
arr = np.array([10, 20, 30, 40, 50, 60])

# Select values greater than 30
print(arr[arr > 30])

# Select values less than 40
print(arr[arr < 40])

# Select even numbers
print(arr[arr % 2 == 0])

# Select odd numbers
print(arr[arr % 2 != 0])

# Select values between 20 and 50
print(arr[(arr >= 20) & (arr <= 50)])

# Select values greater than 20 OR less than 50
print(arr[(arr > 20) | (arr < 50)])
```

conditional function

```python
marks = np.array([35, 45, 80, 25, 90])

# Replace marks below 40 with "Fail", otherwise "Pass"
result = np.where(marks >= 40, "Pass", "Fail")

print(result)

# Clip values between 30 and 80
print(np.clip(marks, 30, 80))

# Check whether all values are greater than zero
print(np.all(marks > 0))

# Check whether any value is greater than 85
print(np.any(marks > 85))
```

random number generation 

```python
import numpy as np

# Generate one random number between 0 and 1
print(np.random.rand())

# Generate a 2x3 random array
print(np.random.rand(2, 3))

# Generate random integers between 1 and 100
print(np.random.randint(1, 101, size=5))

# Generate random numbers from a normal distribution
print(np.random.randn(5))

# Shuffle array elements in place
arr = np.array([10, 20, 30, 40, 50])
np.random.shuffle(arr)
print(arr)

# Select 3 random elements without replacement
print(np.random.choice([10, 20, 30, 40, 50], 3, replace=False))

# Set seed for reproducible random results
np.random.seed(42)
print(np.random.randint(1, 100, 5))
```

copy vs view

```python
arr = np.array([10, 20, 30, 40])

# Create an independent copy
copy_arr = arr.copy()

# Modify the copied array
copy_arr[0] = 100

# Original array remains unchanged
print(arr)

# Create a view of the original array
view_arr = arr.view()

# Modify the view
view_arr[1] = 200

# Original array also changes
print(arr)
```

Broadcasting

```python
# Create an array
arr = np.array([10, 20, 30])

# Add 5 to every element
print(arr + 5)

# Multiply every element by 2
print(arr * 2)

# Broadcasting with two-dimensional array
a = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

b = np.array([10, 20, 30])

# Add b to every row of a
print(a + b)
```

matrix operations

```python
a = np.array([
    [1, 2],
    [3, 4]
])

b = np.array([
    [5, 6],
    [7, 8]
])

# Matrix addition
print(a + b)

# Matrix multiplication (element-wise)
print(a * b)

# Matrix multiplication (dot product)
print(np.dot(a, b))

# Alternative matrix multiplication operator
print(a @ b)

# Transpose matrix
print(a.T)

# Calculate determinant
print(np.linalg.det(a))

# Calculate inverse
print(np.linalg.inv(a))

# Calculate eigenvalues
print(np.linalg.eigvals(a))
```

handling missing values (NaN)

```python
arr = np.array([10, 20, np.nan, 40, 50])

# Check which values are NaN
print(np.isnan(arr))

# Calculate mean while ignoring NaN
print(np.nanmean(arr))

# Calculate sum while ignoring NaN
print(np.nansum(arr))

# Calculate maximum while ignoring NaN
print(np.nanmax(arr))

# Replace NaN with zero
arr[np.isnan(arr)] = 0

print(arr)
```

unique values and counting

```python
arr = np.array([10, 20, 10, 30, 20, 40, 10])

# Find unique values
print(np.unique(arr))

# Find unique values and their frequency
values, counts = np.unique(arr, return_counts=True)

print("Values:", values)
print("Counts:", counts)
```

Advanced NumPy functions

```python
arr = np.array([10, 20, 30, 40, 50])

# Change data type to float
print(arr.astype(float))

# Check whether array elements are finite
print(np.isfinite(arr))

# Check whether values are positive
print(np.sign(arr))

# Calculate difference between consecutive elements
print(np.diff(arr))

# Repeat each element two times
print(np.repeat(arr, 2))

# Repeat the complete array two times
print(np.tile(arr, 2))

# Insert a value at a specified index
print(np.insert(arr, 2, 99))

# Delete an element by index
print(np.delete(arr, 2))

# Append an element
print(np.append(arr, 60))
```

Practical DAta Analytics Project Using Numpy 

```python
import numpy as np

# Employee names
names = np.array([
    "Rahul",
    "Aman",
    "Vikas",
    "Priya",
    "Rohit"
])

# Employee salaries
salary = np.array([
    25000,
    35000,
    45000,
    30000,
    50000
])

# Display employee names and salaries
print("Employee Names:", names)
print("Employee Salaries:", salary)

# Calculate total salary expense
print("Total Salary:", np.sum(salary))

# Calculate average salary
print("Average Salary:", np.mean(salary))

# Find highest salary
print("Highest Salary:", np.max(salary))

# Find lowest salary
print("Lowest Salary:", np.min(salary))

# Find employee with highest salary
print("Highest Paid Employee:", names[np.argmax(salary)])

# Find employees earning more than 30000
print("Employees Above 30000:", names[salary > 30000])

# Increase every employee's salary by 10%
new_salary = salary * 1.10

print("Updated Salaries:", new_salary)

# Sort salaries
print("Sorted Salaries:", np.sort(salary))
```
