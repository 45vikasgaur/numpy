```python
import numpy as np

a = np.array([10, 20, 30])
b = np.array([2, 4, 5])

# Addition
print(a + b)

# Subtraction
print(a - b)

# Multiplication
print(a * b)

# Division
print(a / b)

# Power
print(a ** 2)

# Modulus
print(a % b)
```

```python
arr = np.array([10, 20, 30, 40, 50])

print(arr > 25)
print(arr < 30)
print(arr == 20)
print(arr != 40)
print(arr >= 30)
print(arr <= 20)
```

```python
arr = np.array([10, 20, 30, 40, 50, 60])

# Greater than 20 AND less than 50
print(arr[(arr > 20) & (arr < 50)])

# Less than 20 OR greater than 50
print(arr[(arr < 20) | (arr > 50)])

# NOT condition
print(arr[~(arr > 30)])
```
