```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print(arr[0])  # First element
print(arr[1])  # Second element
print(arr[4])  # Fifth element
```

```python
arr = np.array([10, 20, 30, 40, 50])

print(arr[-1])  # Last element
print(arr[-2])  # Second last element
```

```python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(arr[0, 0])  # 10
print(arr[1, 2])  # 60
print(arr[2, 1])  # 80
```

```python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

# First row
print(arr[0, :])

# Second column
print(arr[:, 1])

# Last row
print(arr[-1, :])
```
