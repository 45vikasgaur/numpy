```python
import numpy as np

arr = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print("Array:", arr)

# Number of dimensions
print("Dimensions:", arr.ndim)

# Rows and columns
print("Shape:", arr.shape)

# Total elements
print("Size:", arr.size)

# Data type
print("Data Type:", arr.dtype)

# Memory used by one element
print("Item Size:", arr.itemsize)

# Total memory used
print("Total Bytes:", arr.nbytes)
```
