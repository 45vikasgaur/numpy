NUMPY INDEXING AND SLICING OF MATRIX

```python
arr = np.arange(1,31).reshape(6,5)
arr
```

```python
slice = arr[0:2, 1:3]
slice
```

```python
slice = arr[3:,3:]
slice
```

```python
arr[:,4]
```
