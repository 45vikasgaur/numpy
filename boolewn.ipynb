```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50, 60])

# Select values greater than 30
result = arr[arr > 30]

print(result)
```

```python
arr = np.array([10, 20, 30, 40, 50, 60])

# Values less than 40
print(arr[arr < 40])

# Values equal to 30
print(arr[arr == 30])

# Even numbers
print(arr[arr % 2 == 0])

# Odd numbers
print(arr[arr % 2 != 0])
```

```python
arr = np.array([10, 20, 30, 40, 50, 60])

# Select indexes 0, 2 and 4
print(arr[[0, 2, 4]])
```

```python
arr = np.array([
    [10, 20],
    [30, 40],
    [50, 60]
])

print(arr[[0, 2]])
```
