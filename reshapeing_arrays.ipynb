```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5, 6])

# Convert 1D into 2D
new_arr = arr.reshape(2, 3)

print(new_arr)
```

```python
arr = np.array([1, 2, 3, 4, 5, 6])

print(arr.reshape(3, -1))
```

```python
arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

new_arr = arr.flatten()

print(new_arr)
```

```python
arr = np.array([
    [10, 20],
    [30, 40]
])

new_arr = arr.ravel()

print(new_arr)
```

```python
arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(arr.T)
```

```python
arr = np.array([1, 2, 3, 4])

new_arr = np.resize(arr, (3, 3))

print(new_arr)
```

```python
arr = np.array([10, 20, 30])

new_arr = np.expand_dims(arr, axis=0)

print(new_arr)
print(new_arr.shape)
```
