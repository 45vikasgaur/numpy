```python
import numpy as np

arr = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print("Array:", arr)

# Number of dimensions
print("Dimensions:", arr.ndim)

# Rows and columns
print("Shape:", arr.shape)

# Total elements
print("Size:", arr.size)

# Data type
print("Data Type:", arr.dtype)

# Memory used by one element
print("Item Size:", arr.itemsize)

# Total memory used
print("Total Bytes:", arr.nbytes)
```

```python
import numpy as np

# Integer array
a = np.array([10, 20, 30], dtype=np.int32)

# Float array
b = np.array([10.5, 20.5, 30.5], dtype=np.float64)

# Boolean array
c = np.array([True, False, True])

print(a.dtype)
print(b.dtype)
print(c.dtype)
```

```python
import numpy as np

arr = np.array([10, 20, 30, 40])

# Convert integer to float
new_arr = arr.astype(float)

print(new_arr)
print(new_arr.dtype)
```
